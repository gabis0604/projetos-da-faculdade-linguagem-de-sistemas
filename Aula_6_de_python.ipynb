```python
convidados = ("Alice", "Rodrigo", "Carol", "felipe")
confirmados = ["Alice", "Rodrigo"]
```

```python
print("Convidados que ainda não confirmaram: ")
for convidado in convidados:
  if convidado not in confirmados:
    print(f"- {convidado}")
```
