```python
from IPython.display import display, HTML

html = """
<div style="display:flex; gap:40px; align-items:flex-start;
            font-family:Arial; flex-wrap:wrap;">

    <!-- CALCULADORA -->
    <div>
        <p style="font-size:14px; color:#777;">
            📱 simulação do app KivyMD
        </p>

        <div style="
            width:280px;
            background:#f5f5f5;
            border:1px solid #ddd;
            border-radius:25px;
            padding:20px;
        ">

            <!-- DISPLAY -->
            <div style="
                background:white;
                border:1px solid #ccc;
                border-radius:10px;
                padding:12px;
                margin-bottom:15px;
            ">

                <div style="font-size:12px; color:#aaa;">
                    MDTextField
                </div>

                <div id="expr" style="
                    font-size:22px;
                    text-align:right;
                    min-height:30px;
                "></div>

            </div>


            <!-- BOTÕES -->
            <div class="calc-grid">

                <button class="num" onclick="pressNum('7')">7</button>
                <button class="num" onclick="pressNum('8')">8</button>
                <button class="num" onclick="pressNum('9')">9</button>
                <button class="op" onclick="pressOp('*')">×</button>

                <button class="num" onclick="pressNum('4')">4</button>
                <button class="num" onclick="pressNum('5')">5</button>
                <button class="num" onclick="pressNum('6')">6</button>
                <button class="op" onclick="pressOp('/')">÷</button>

                <button class="num" onclick="pressNum('1')">1</button>
                <button class="num" onclick="pressNum('2')">2</button>
                <button class="num" onclick="pressNum('3')">3</button>
                <button class="op" onclick="pressOp('+')">+</button>

                <button class="clear" onclick="clearAll()">C</button>
                <button class="num" onclick="pressNum('0')">0</button>
                <button class="equal" onclick="calculate()">=</button>
                <button class="op" onclick="pressOp('-')">−</button>

            </div>

            <div style="
                text-align:center;
                margin-top:15px;
                font-size:11px;
                color:#aaa;
            ">
                KivyMD · Material Design
            </div>

        </div>
    </div>


    <!-- PARTE DIREITA -->
    <div style="flex:1; min-width:350px;">

        <p style="font-size:14px; color:#777;">
            ♻️ método chamado no Python
        </p>

        <div id="code-panel" style="
            background:#f8f8f8;
            border:1px solid #ddd;
            border-radius:12px;
            padding:15px;
            font-family:monospace;
            font-size:13px;
            min-height:100px;
        ">
            <span style="color:#aaa;">
                ← clique em um botão
            </span>
        </div>


        <p style="
            font-size:14px;
            color:#777;
            margin-top:18px;
        ">
            📋 histórico de operações
        </p>

        <div id="history" style="
            background:#f8f8f8;
            border:1px solid #ddd;
            border-radius:10px;
            padding:15px;
            min-height:50px;
            font-size:13px;
            color:#666;
        ">
            <span style="color:#aaa;">
                nenhuma operação ainda
            </span>
        </div>

        <div id="description"
             style="margin-top:18px; font-size:14px;">
        </div>

    </div>

</div>


<style>

.calc-grid {
    display:grid;
    grid-template-columns:repeat(4, 1fr);
    gap:8px;
}

.calc-grid button {
    height:55px;
    border-radius:10px;
    border:1px solid #ccc;
    font-size:17px;
    cursor:pointer;
}

/* números */
.num {
    background:white;
}

/* operadores */
.op {
    background:#d9edf7;
}

/* botão limpar */
.clear {
    background:#f6b79c;
}

/* botão igual */
.equal {
    background:#d7f0d2;
}

.calc-grid button:hover {
    opacity:0.8;
}

</style>


<script>

let expression = "";
let historyList = [];

function updateDisplay() {
    document.getElementById("expr").innerText = expression;
}


function pressNum(num) {

    expression += num;
    updateDisplay();

    document.getElementById("code-panel").innerHTML =
    "def add_number(self, number):<br>" +
    "&nbsp;&nbsp;&nbsp;&nbsp;self.ids.input_field.text += number";

    document.getElementById("description").innerHTML =
    "<b>add_number()</b> — adiciona o número selecionado.";
}


function pressOp(op) {

    expression += op;
    updateDisplay();

    document.getElementById("code-panel").innerHTML =
    "def add_operator(self, operator):<br>" +
    "&nbsp;&nbsp;&nbsp;&nbsp;self.ids.input_field.text += operator";

    document.getElementById("description").innerHTML =
    "<b>add_operator()</b> — adiciona o operador selecionado.";
}


function clearAll() {

    expression = "";
    updateDisplay();

    document.getElementById("code-panel").innerHTML =
    "def clear(self):<br>" +
    "&nbsp;&nbsp;&nbsp;&nbsp;self.ids.input_field.text = ''";

    document.getElementById("description").innerHTML =
    "<b>clear()</b> — limpa o conteúdo da calculadora.";
}


function calculate() {

    try {

        let original = expression;

        let result = eval(expression);

        expression = String(result);

        updateDisplay();

        historyList.unshift(original + " = " + result);

        document.getElementById("history").innerHTML =
            historyList.join("<br><br>");


        document.getElementById("code-panel").innerHTML =
        "def calculate_result(self):<br>" +
        "&nbsp;&nbsp;&nbsp;&nbsp;try:<br>" +
        "&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;result = eval(self.ids.input_field.text)<br>" +
        "&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.ids.input_field.text = str(result)<br>" +
        "&nbsp;&nbsp;&nbsp;&nbsp;except Exception as e:<br>" +
        "&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.ids.input_field.text = 'Erro'";


        document.getElementById("description").innerHTML =
        "<b>calculate_result()</b> — avalia com eval(). Resultado: <b>" +
        result + "</b>";

    }

    catch {

        expression = "Erro";
        updateDisplay();

    }

}

</script>
"""

display(HTML(html))
```
