```python
import sqlite3

conn = sqlite3.connect('minhaloja.db')
cursor = conn.cursor()
```

```python
create_table = """
CREATE TABLE IF NOT EXISTS Produtos (
  id INTEGER PRIMARY KEY,
  nome TEXT NOT NULL,
  preco REAL NOT NULL,
  estoque INTEGER)
"""

```

```python
cursor.execute(create_table)
conn.commit()
conn.close()
print("tabela produtos criada.")
```

```python
import sqlite3

conn = sqlite3.connect('minhaloja.db')
cursor = conn.cursor()

novo_preco = 27.99
produto_id = 3

cursor.execute(
   "UPDATE Produtos SET preco = ? WHERE id = ?",
    (novo_preco, produto_id)
)

conn.commit()
conn.close()
print("Preço do produto atualizado")
```

```python
import sqlite3

conn = sqlite3.connect('contatos.db')
cursor = conn.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS Contatos (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nome TEXT,
        telefone TEXT,
        email TEXT
    )
""")

dados_exemplos= [
    ('João Silva', '123-456-7890', 'william.henry.harrison@example-pet-store.com'),
    ('Maria Santos', '9644-554-941', 'maria.@ninguem.com'),
    ('José Pereira', '544-582-597', 'jose_125@exemplo.com'),
]
cursor.executemany(
    "INSERT INTO Contatos (nome, telefone, email) VALUES (?, ?, ?)",
    dados_exemplos
)

conn.commit()

cursor.execute("SELECT * FROM Contatos")
contatos = cursor.fetchall()
print("Contatos:")
for contato in contatos:
    print(contato)

cursor.execute(
    "UPDATE Contatos SET telefone = ? WHERE id = ?",
    ('123-456-789', 2)
)
conn.commit()

cursor.execute("DELETE FROM Contatos WHERE id = ?", (1,))
conn.commit()

cursor.execute("SELECT * FROM Contatos")
print("\nContatos após UPDATE e DELETE:")
for contato in cursor.fetchall(): # Corrected typo here
    print(contato)

cursor.execute("SELECT COUNT(DISTINCT email) FROM Contatos")
quantidade_emails = cursor.fetchone()[0]

print("\nQuantidade de e-mails distintos:", quantidade_emails)


conn.close()
```
