```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]

plt.plot(x, y)

plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Gráfico de Linhas')

plt.show()
```

```python
import matplotlib.pyplot as plt

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Agos', 'Set', 'Out', 'Nov', 'Dez']
vendas = [1120, 1090, 1150, 1280, 1200, 1340, 2110, 2180, 2160, 3530, 3550, 4500]

plt.bar(meses, vendas, color='royalblue')

plt.xlabel('Mês')
plt.ylabel('Vendas (em unidades)')
plt.title('Vendas Mensais')

plt.show()
```
