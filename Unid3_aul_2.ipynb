```python
exemplo2 = {'a': 10, 'b': 20, 'c': 30, 'd': 40, 'e': 50}
series2 = pd.Series(data=exemplo2)
print(series2)
```

```python
print("Media:", series2.mean())
print("Soma:", series2.sum())
print("Maximo:", series2.max())
print("Minimo:", series2.min())
```

```python
import pandas as pd

dados = {
    'Nome': ['Alice', 'Bob', 'Charlie', 'David'],
    'Idade': [25, 30, 35, 40],
}

serie_idades = pd.Series(dados['Idade'], index=dados['Nome'])

print("Serie de Idades: ")
print(serie_idades)

media_idades = serie_idades.mean()
print(f"\nMedia das Idades: , {media_idades}")

print(f"\nMaior Idade: {serie_idades.max()}")
print(f"Menor Idade: {serie_idades.min()}")
```
