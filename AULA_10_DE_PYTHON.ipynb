```python
import matplotlib.pyplot as plt
```

```python
class Produto:
  def __init__(self, nome, preco, categoria, estoque):
    self.nome = nome
    self.preco = preco
    self.categoria = categoria
    self.estoque = estoque

  def __str__(self):
    return (f"{self.nome} | R$ {self.preco:.2f} | "
             f"Categoria: {self.categoria} | Estoque: {self.estoque}")

```

```python
catalago = []
categorias = []
```

```python
def adicionar_produto(nome, preco, categoria, estoque):
  novo_produto = Produto(nome, preco, categoria, estoque)
  catalago.append(novo_produto)
  categorias.append(categoria)
  print(f"O produto {nome} adicionado ao catálogo!")
```

```python
def listar_catalogo():
  print("\n---Catálogo de Produtos da minha loja---")
  for produto in catalago:
    print(produto)
```

```python
adicionar_produto("Notebook",  3550.00, "Informatica", 10)
adicionar_produto("Iphone",  10000.00, "Celular", 5)
adicionar_produto("Cafeteira", 350.00, "Eletrodoméstico", 2)
adicionar_produto("Geladeira", 4500.00, "Eletrodoméstico", 8)
adicionar_produto("Fone Bluetooth", 200.00, "Acessórios", 15)

listar_catalogo()

```

```python
categorias_unicas = list(set(categorias))
categorias_unicas.sort()

contagem = [categorias.count(cat) for cat in categorias_unicas]

plt.figure(figsize=(9, 5))

barras = plt.bar(
    categorias_unicas,
    contagem,
    color="royalblue",
    edgecolor="black"
)

plt.xlabel("Categorias")
plt.ylabel("Quantidade")
plt.title("Quantidade de Produtos por Categoria")

for barra, valor in zip(barras, contagem):
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + 0.05,
        str(valor),
        ha="center"
    )

plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

```
